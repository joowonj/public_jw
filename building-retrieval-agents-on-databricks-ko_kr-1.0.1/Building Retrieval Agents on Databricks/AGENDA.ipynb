## Databricks에서 Retrieval Agents 구축하기

이 과정은 Databricks Data Intelligence Platform에서 Retrieval Agents를 구축하는 실습 교육을 제공합니다. 참가자들은 비정형 문서를 구조화된 데이터로 파싱하고, 검색 워크플로를 위한 콘텐츠 변환 및 청크 생성, 문서 검색을 위한 AI Search 솔루션 구축, MLflow 및 Agent Bricks를 활용한 운영 준비 에이전트 개발을 배우게 됩니다. 이 과정은 문서 처리부터 임베딩 생성, 벡터 인덱싱, 거버넌스 기능을 갖춘 에이전트 배포까지 에이전트의 전체 라이프사이클을 다룹니다.

---

## 선수 조건

- 중급 Python 프로그래밍 경험
- 쿼리 및 함수 생성을 위한 기본 SQL 지식
- Databricks Data Intelligence Platform에 대한 친숙함
- 카탈로그와 스키마를 포함한 Unity Catalog 개념 이해
- 대형 언어 모델(LLM) 및 프롬프트 엔지니어링에 대한 기본 이해
- 실험 추적을 위한 MLflow 기본 지식
